```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/Healthcare Analytics for Doctor Visits.csv")
```

```python
df['age']=df['age']*100
df['income']=df['income']*100
```

```python
df=df.astype({'age':int, 'income':int})
```

```python
df=df.drop('Unnamed: 0',axis=1)
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='gender',data=df, palette='viridis', hue='gender')
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df['age'],bins=20,kde=True)
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df['visits'],bins=range(df['visits'].max()+2),color='maroon')
plt.title("Distribution of Doctor Visits")
plt.xlabel("Visits")
plt.ylabel("Count")
plt.xticks(range(df['visits'].max()+1))
plt.show()
```

```python
gender_analysis=(df.groupby('gender')['visits'].agg(['count','mean','median']).reset_index())
display(gender_analysis)
sns.barplot(data=gender_analysis,x="gender",y="mean",hue="gender")
plt.title("Average Doctor Visits by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Visits")
plt.show()
```

```python
illness_analysis=(df.groupby('illness')['visits'].agg(['count','mean']).reset_index())
display(illness_analysis)
plt.figure(figsize=(6,4))
sns.lineplot(data=illness_analysis,x='illness',y='mean',marker='o')
plt.title("Average Doctor Visits by Illness")
plt.xlabel("Illness")
plt.ylabel("Average Visits")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.boxplot(data=df,x="gender",y="visits")
plt.title("Doctor Visits by Gender")
plt.xlabel("Gender")
plt.ylabel("Visits")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x="age",y="visits",hue="gender")
plt.title("Doctor Visits by Age")
plt.xlabel("Age")
plt.ylabel("Visits")
plt.show()
```
